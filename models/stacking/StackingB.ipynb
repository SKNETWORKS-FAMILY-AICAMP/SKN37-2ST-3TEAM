## 1 데이터 불러오기

```python
import numpy as np
import pandas as pd 
```

```python
df=pd.read_csv(
    'hotel_preprocessed.csv',
    dtype={"agent": str})

display(df.columns)
```

## 2 데이터 분할

### 2.1 Train/Test 분리 

```python
ID_COLS = ["source_row_id", "group_id", "split"]
TARGET = "is_canceled"

FEATURES = [
    col for col in df.columns
    if col not in ID_COLS + [TARGET]
]

train = df[df["split"] == "train"].copy()
test = df[df["split"] == "test"].copy()

X_train_all = train[FEATURES]
y_train_all = train[TARGET]
groups_train_all = train["group_id"]

X_test = test[FEATURES]
y_test = test[TARGET]

print("Train:", X_train_all.shape)
print("Test:", X_test.shape)
```

### 2.2 Train 내부 Validation 분리 

```python
from sklearn.model_selection import StratifiedGroupKFold

sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=0
)

train_idx, valid_idx = next(
    sgkf.split(
        X_train_all,
        y_train_all,
        groups_train_all
    )
)

X_train = X_train_all.iloc[train_idx].copy()
y_train = y_train_all.iloc[train_idx].copy()
groups_train = groups_train_all.iloc[train_idx].copy()

X_valid = X_train_all.iloc[valid_idx].copy()
y_valid = y_train_all.iloc[valid_idx].copy()

print("Train:", len(X_train))
print("Validation:", len(X_valid))
print("Test:", len(X_test))
```

### 2.3 그룹 중복 확인

```python
train_groups = set(groups_train)
valid_groups = set(
    groups_train_all.iloc[valid_idx]
)
test_groups = set(test["group_id"])

print("Train / Valid:", len(train_groups & valid_groups))
print("Train / Test:", len(train_groups & test_groups))
print("Valid / Test:", len(valid_groups & test_groups))
```

## 3 모델별 전처리 Pipeline 생성

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

# 수치형 컬럼
numeric_cols = X_train.select_dtypes(
    include=np.number
).columns.tolist()

# 범주형 컬럼
categorical_cols = X_train.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

# 트리 모델용 수치형 전처리
numeric_tree = "passthrough"

# 로지스틱 회귀용 수치형 전처리
numeric_linear = Pipeline([
    ("scaler", StandardScaler())
])

# 범주형 전처리
categorical_pipe = Pipeline([
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        min_frequency=100
    ))
])

# XGBoost / Random Forest
tree_preprocessor = ColumnTransformer([
    ("num", numeric_tree, numeric_cols),
    ("cat", categorical_pipe, categorical_cols)
])

# Logistic Regression
linear_preprocessor = ColumnTransformer([
    ("num", numeric_linear, numeric_cols),
    ("cat", categorical_pipe, categorical_cols)
])
```

## 4 Base Model 정의 

```python
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier

models = {
    "XGBoost": Pipeline([
        ("preprocessor", tree_preprocessor),
        ("model", XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=0.05,
            eval_metric="logloss",
            random_state=42,
            n_jobs=2
        ))
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", tree_preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=200,
            max_depth=15,
            min_samples_leaf=3,
            random_state=42,
            n_jobs=2
        ))
    ]),

    "MLP": Pipeline([
        ("preprocessor", linear_preprocessor),
        ("model", MLPClassifier(
            hidden_layer_sizes=(64, 32),
            activation="relu",
            solver="adam",
            alpha=0.001,
            learning_rate_init=0.001,
            batch_size=256,
            max_iter=300,
            early_stopping=True,
            n_iter_no_change=15,
            random_state=42
        ))
    ])
}
```

### RandomizedSearchCV

```python
param_distributions = {
    "XGBoost": {
        "model__n_estimators": [200, 300, 500],
        "model__max_depth": [3, 4, 5],
        "model__learning_rate": [0.03, 0.05, 0.1],
        "model__min_child_weight": [1, 3, 5],
        "model__subsample": [0.8, 1.0],
        "model__colsample_bytree": [0.8, 1.0],
        "model__reg_lambda": [1, 5, 10]
    },
    "Random Forest": {
        "model__n_estimators": [200, 300, 500],
        "model__max_depth": [10, 15, 20],
        "model__min_samples_leaf": [3, 5, 10, 20],
        "model__min_samples_split": [2, 10, 20],
        "model__max_features": ["sqrt", 0.5]
    },
    "MLP": {
        "model__hidden_layer_sizes": [
            (32,),
            (64, 32),
            (128, 64),
            (128, 64, 32)
        ],
        "model__alpha": [0.0001, 0.001, 0.01],
        "model__learning_rate_init": [0.0005, 0.001, 0.005],
        "model__batch_size": [128, 256]
    }
}
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedGroupKFold

tuning_cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

best_models = {}
tuning_results = []

for name, model in models.items():

    print(f"\n{name} 하이퍼파라미터 튜닝 시작")

    search = RandomizedSearchCV(
        estimator=model,
        param_distributions=param_distributions[name],
        n_iter=20,
        scoring="roc_auc",
        cv=tuning_cv,
        n_jobs=-1,
        random_state=42,
        refit=True,
        verbose=1
    )

    search.fit(
        X_train,
        y_train,
        groups=groups_train
    )

    best_models[name] = search.best_estimator_

    tuning_results.append({
        "Model": name,
        "Best ROC-AUC": search.best_score_,
        "Best Params": search.best_params_
    })

    print("Best ROC-AUC:", search.best_score_)
    print("Best Params:", search.best_params_)

tuning_results_df = pd.DataFrame(tuning_results)
display(tuning_results_df)
```

## 5 그룹 기반 OOF 예측 생성

```python
from sklearn.base import clone
from sklearn.model_selection import StratifiedGroupKFold

cv = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# OOF 예측을 저장할 DataFrame
oof_pred = pd.DataFrame(
    index=np.arange(len(X_train)),
    columns=list(models.keys()),
    dtype=float
)

# 실제 반복 학습 
fitted_models = {}

for name, model in best_models.items():

    print(f"\n{name} OOF 학습 시작")

    for fold, (tr_idx, val_idx) in enumerate(
        cv.split(X_train, y_train, groups_train),
        start=1
    ):
        fold_model = clone(model) 

        fold_model.fit(
            X_train.iloc[tr_idx],
            y_train.iloc[tr_idx]
        )

        probabilities = fold_model.predict_proba(
            X_train.iloc[val_idx]
        )[:, 1]

        oof_pred.iloc[
            val_idx,
            oof_pred.columns.get_loc(name)
        ] = probabilities

        print(f"Fold {fold} 완료")

    # Validation/Test 예측용 전체 Train 학습
    final_model = clone(model)
    final_model.fit(X_train, y_train)

    fitted_models[name] = final_model

print("\nOOF 결측치:", oof_pred.isna().sum().sum())
display(oof_pred.head())
```

## 6 Meta model 학습

```python
from sklearn.linear_model import LogisticRegression

meta_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

meta_model.fit(
    oof_pred,
    y_train.to_numpy()
)
```

## 7 모델 성능 및 비교


```python
# Validation 데이터에 대한 Base Model 예측
valid_meta_X = pd.DataFrame({
    name: model.predict_proba(X_valid)[:, 1]
    for name, model in fitted_models.items()
})

# Meta Model로 최종 취소 확률 계산
stacking_valid_proba = meta_model.predict_proba(
    valid_meta_X
)[:, 1]

# 취소 여부 예측 (기준 0.5)
stacking_valid_pred = (
    stacking_valid_proba >= 0.5
).astype(int)
```

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    brier_score_loss
)

print("Accuracy:",
      accuracy_score(y_valid, stacking_valid_pred))

print("F1:",
      f1_score(y_valid, stacking_valid_pred))

print("ROC-AUC:",
      roc_auc_score(y_valid, stacking_valid_proba))

print("PR-AUC:",
      average_precision_score(y_valid, stacking_valid_proba))

print("Brier Score:",
      brier_score_loss(y_valid, stacking_valid_proba))
```

## 8 단일 모델과 stacking 비교

```python
results = []

# 단일 모델 평가
for name, model in fitted_models.items():

    proba = model.predict_proba(X_valid)[:, 1]
    pred = (proba >= 0.5).astype(int)

    results.append({
        "Model": name,
        "ROC-AUC": roc_auc_score(y_valid, proba),
        "PR-AUC": average_precision_score(y_valid, proba),
        "F1": f1_score(y_valid, pred),
        "Brier Score": brier_score_loss(y_valid, proba)
    })

# Stacking 평가
results.append({
    "Model": "Stacking B",
    "ROC-AUC": roc_auc_score(
        y_valid, stacking_valid_proba
    ),
    "PR-AUC": average_precision_score(
        y_valid, stacking_valid_proba
    ),
    "F1": f1_score(
        y_valid, stacking_valid_pred
    ),
    "Brier Score": brier_score_loss(
        y_valid, stacking_valid_proba
    )
})

results_df = pd.DataFrame(results)

display(
    results_df.sort_values(
        "ROC-AUC",
        ascending=False
    )
)
```

## 11 특성중요도

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import roc_auc_score

# Stacking 모델 예측 함수
def stacking_predict_proba(X):
    base_pred = pd.DataFrame({
        name: fitted_models[name].predict_proba(X)[:, 1]
        for name in oof_pred.columns
    }, index=X.index)

    return meta_model.predict_proba(base_pred)[:, 1]


# 기존 Validation 성능
baseline_auc = roc_auc_score(
    y_valid, stacking_predict_proba(X_valid)
)

# 특성 중요도 계산
results = []
rng = np.random.default_rng(42)

for col in X_valid.columns:
    scores = []

    for _ in range(3):
        X_temp = X_valid.copy()

        # 해당 특성의 값을 무작위로 섞기
        X_temp[col] = rng.permutation(X_temp[col].values)

        perm_auc = roc_auc_score(
            y_valid,
            stacking_predict_proba(X_temp)
        )

        scores.append(baseline_auc - perm_auc)

    results.append({
        "Feature": col,
        "Importance": np.mean(scores)
    })

# 중요도 정렬
importance_df = pd.DataFrame(results).sort_values(
    "Importance", ascending=False
)

display(importance_df)
```

```python
import plotly.express as px

# 중요도 상위 15개
top_features = importance_df.head(15).sort_values(
    "Importance", ascending=True
)

fig = px.bar(
    top_features,
    x="Importance",
    y="Feature",
    orientation="h",
    title="Stacking Model - Feature Importance",
    labels={
        "Importance": "Permutation Importance (ROC-AUC 감소량)",
        "Feature": "Feature"
    },
    text="Importance"
)

fig.update_traces(
    texttemplate="%{text:.4f}",
    textposition="outside"
)

fig.update_layout(
    height=600,
    width=950,
    yaxis_title=None,
    xaxis_title="Importance (ROC-AUC 감소량)"
)

fig.show()
```
